```python
def diffence(n, m):
    dif = n - m
    if dif < 0:
        dif = dif * -1
    return dif

s = "abcd"
score = 0

for i in range(1, len(s)):
    n = ord(s[i - 1])
    m = ord(s[i])
    score += diffence(n, m)   # sum, not compare

print(score)
```
