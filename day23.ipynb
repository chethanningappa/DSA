```python
nums = [1,2,3,1,1,3]

pairs = []
for l, k in enumerate(nums):
    for i, j in enumerate(nums):
        if k == j and l < i:
            pairs.append((l, i))

print(len(pairs),pairs)


```

```python
from collections import Counter

nums = [1,2,3,1,1,3]
freq = Counter(nums)

count = 0
for v in freq.values():
    count += v * (v - 1) // 2
```
