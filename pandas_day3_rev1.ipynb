```python
df = pd.DataFrame({
    "department":["IT","IT","HR","HR"],
    "gender":["M","F","M","F"],
    "salary":[50000,80000,60000,55000]
})

df.groupby(["department","gender"])["salary"].mean()
```

```python
df = pd.DataFrame({
    "department":["IT","IT","HR","HR"],
    "gender":["M","F","M","F"],
    "salary":[50000,80000,60000,55000]
})

df.groupby(["department","gender"])["salary"].mean().reset_index()
```

```python
# highest salary of employee per department
df.loc[df.groupby("department")["salary"].idxmax()]
```

```python
# AVG(salary) OVER (PARTITION BY department)
df["avg_salary"] = df.groupby("department")["salary"].transform("mean")
```

```python
# RANK() OVER (PARTITION BY department ORDER BY salary DESC)
df["rank"] = df.groupby("department")["salary"].rank(ascending=False)
```

```python
# RANK() OVER (PARTITION BY department ORDER BY salary DESC)
```

```python
df['rank']=df.groupby("department")["salary"].rank(ascending=False)
df
```

```python
# SUM(salary) OVER (PARTITION BY department ORDER BY id)
df["running_total"] = df.groupby("department")["salary"].cumsum()
df
```

```python
# LAG(salary)
```

```python
df["prev_salary"] = df.groupby("department")["salary"].shift(1)
df
```

```python
#time series data
```

```python
df["moving_avg"] = df["salary"].rolling(2).mean()
```
